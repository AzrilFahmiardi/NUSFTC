## 1.1 Twitter / X Scraping

## 1.4 Penggabungan Data (Merge)

Setelah semua data terkumpul (baik saat ini juga atau dari run sebelumnya), jalankan sel di bawah ini untuk menggabungkan semuanya menjadi satu dataset utuh.

```python
import os
import pandas as pd
from config.settings import DATA_RAW

def merge_raw_data():
    dfs = []
    file_paths = [
        DATA_RAW / "twitter_raw.csv",
        # DATA_RAW / "tokopedia_raw.csv",
        # DATA_RAW / "shopee_raw.csv"
    ]
    
    for fp in file_paths:
        if fp.exists():
            print(f"Loading {fp.name}...")
            df_tmp = pd.read_csv(fp)
            dfs.append(df_tmp)
        else:
            print(f"File {fp.name} belum ada, dilewati.")
            
    if not dfs:
        print("Belum ada data raw sama sekali. Silakan jalankan scraper di atas.")
        return None
        
    df_all = pd.concat(dfs, ignore_index=True)
    
    # Standardisasi nama kolom teks (karena twitter punya 'text' tapi e-commerce juga 'text')
    # Pastikan minimal ada kolom 'text' dan 'source'
    # Jika sumber dari twitter, tambahkan default 'source'
    if 'source' not in df_all.columns:
        df_all['source'] = 'twitter'  # Fallback kalau belum diset
        
    # Fillna untuk source twitter dari e-commerce yg kosong
    df_all.loc[df_all['source'].isna(), 'source'] = 'twitter'
    
    output_path = DATA_RAW / "all_data_merged.csv"
    df_all.to_csv(output_path, index=False)
    print(f"\nData digabung! Total: {len(df_all)} baris.")
    print(f"Tersimpan di: {output_path}")
    
    return df_all

df_merged = merge_raw_data()
if df_merged is not None:
    print("\nDistribusi sumber data:")
    print(df_merged['source'].value_counts())
```
