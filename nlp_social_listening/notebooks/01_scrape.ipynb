# 01 · Scrape Twitter

Pull tweets across all 6 query groups (Q1–Q6 mapped). Saves `data/raw/twitter_raw.csv` + scraping log.

**Prerequisites**: `cookies.json` valid. If stale, run `python -m scrapers.twitter_auth` first.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

import asyncio
from scrapers.twitter_scraper import TwitterScraper
from config.queries import QUERY_GROUPS
from config.settings import TWITTER_MAX_PER_QUERY

print('Query groups:')
for g, qs in QUERY_GROUPS.items():
    print(f'  {g:12s}  {len(qs)} queries')
```

```python
# Smoke test: 1 query, limit 20
scraper = TwitterScraper(headless=True)
df_test = await scraper.scrape_and_save(
    filename='twitter_smoke.csv',
    groups={'flavor': ['matcha protein drink']},
    max_per_query=20,
)
print(f'Smoke: {len(df_test)} tweets')
df_test.head()
```

```python
# Full run — all 6 groups
scraper = TwitterScraper(headless=True)
df = await scraper.scrape_and_save(
    filename='twitter_raw.csv',
    groups=QUERY_GROUPS,
    max_per_query=TWITTER_MAX_PER_QUERY,
)
print(f'Total raw tweets: {len(df)}')
df.groupby('query_group').size()
```
