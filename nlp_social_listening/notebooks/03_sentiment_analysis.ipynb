# 3. Sentiment Analysis

Notebook ini menjalankan Dual-Engine Sentiment Analysis (VADER + TextBlob) pada data ulasan yang telah dibersihkan.

```python
import sys
from pathlib import Path
import pandas as pd

base_dir = Path.cwd().parent
sys.path.insert(0, str(base_dir))

from config.settings import DATA_PROCESSED
from analysis.sentiment_analyzer import SentimentAnalyzer

pd.set_option('display.max_colwidth', 150)
```

## 3.1 Load Processed Data

```python
processed_path = DATA_PROCESSED / "cleaned_data.csv"

if not processed_path.exists():
    print(f"File {processed_path} tidak ditemukan. Harap jalankan Notebook 02 terlebih dahulu.")
else:
    df_clean = pd.read_csv(processed_path)
    print(f"Loaded {len(df_clean)} rows.")
```

## 3.2 Run Sentiment Analysis

Menghitung sentimen (positif, netral, negatif) dan skor konfidensi.

```python
if 'df_clean' in locals():
    analyzer = SentimentAnalyzer()
    print("Menjalankan analisis sentimen...")
    df_sentiment = analyzer.analyze_dataframe(df_clean, text_col="clean_text")
    
    print("\nSelesai!")
    # Tampilkan summary statistics
    stats = analyzer.get_summary_stats(df_sentiment)
    import json
    print(json.dumps(stats, indent=2))
```

```python
if 'df_sentiment' in locals():
    # Lihat sampel ulasan yang paling negatif
    print("\n=== Top 5 Ulasan Paling Negatif ===")
    display(df_sentiment.sort_values("vader_compound")[["text", "consensus_label", "vader_compound"]].head(5))
    
    # Lihat sampel ulasan yang paling positif
    print("\n=== Top 5 Ulasan Paling Positif ===")
    display(df_sentiment.sort_values("vader_compound", ascending=False)[["text", "consensus_label", "vader_compound"]].head(5))
```

## 3.3 Save Data with Sentiment

```python
if 'df_sentiment' in locals():
    sentiment_out_path = DATA_PROCESSED / "sentiment_data.csv"
    df_sentiment.to_csv(sentiment_out_path, index=False)
    print(f"Data dengan sentimen disimpan ke: {sentiment_out_path}")
```
